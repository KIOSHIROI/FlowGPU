# FlowGPU in five minutes

This notebook compiles a tiny dataflow program, inspects the generated Triton and runs the portable reference backend. No GPU is required.

Run it from the repository root with JupyterLab or VS Code.


## 1. Look at the program


```python
from pathlib import Path
print(Path('examples/quickstart/add.fgpu').read_text())
```

## 2. Compile and inspect the typed pipeline


```python
from compiler import compile_source
source = Path('examples/quickstart/add.fgpu').read_text()
compiled = compile_source(source)
pipeline = compiled.pipelines[0]
print(pipeline.pipeline)
print('kernels:', [k.name for k in pipeline.triton.kernels])
```

## 3. Read the generated Triton


```python
print(pipeline.triton.kernels[0].source)
print('--- host wrapper ---')
print(pipeline.triton.host)
```

## 4. Run the reference backend

The command below executes the same semantics in portable Python and prints the output values.


```python
!python -m compiler.cli examples/quickstart/add.fgpu --run
```
